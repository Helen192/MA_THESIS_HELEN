```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MaxNLocator

import json
import pickle
import torch
```

# This script is to convert data to latex form

### Merging datasets

```python
model_24_10_result = pd.read_csv('model_24_10_result.csv', index_col=0)
fomo_result = pd.read_csv('default_fomo_result.csv', index_col=0)

bagging_ensemble_55_result = pd.read_csv('bagging_ensemble_55_result.csv', index_col=0)
bagging_ensemble_45_result = pd.read_csv('bagging_ensemble_45_result.csv', index_col=0)
bagging_ensemble_13_8_result = pd.read_csv('bagging_ensemble_13_8_result.csv', index_col=0)

selected_bagging_result = pd.read_csv('selected_bagging_ensemble_13_7_result.csv', index_col=0)
```

```python
datasets = model_24_10_result['dataset']
data = {'datasets': datasets, 'FoMo-0D': fomo_result['aucroc'] * 100, '1st base model': model_24_10_result['aucroc'] * 100, 
        '1st Bagging': bagging_ensemble_55_result['aucroc'] * 100, '2nd Bagging': bagging_ensemble_45_result['aucroc'] * 100, 
        '3rd Bagging': bagging_ensemble_13_8_result['aucroc'] * 100, '1st Selected Bagging': selected_bagging_result['aucroc'] * 100}
df = pd.DataFrame(data)
```

```python
print(df.to_latex(index=False,
                  float_format="{:.2f}".format, longtable=True
                  ))
```

```python
datasets = model_24_10_result['dataset']
data_aucpr = {'datasets': datasets, 'FoMo-0D': fomo_result['aucpr'] * 100, '1st base model': model_24_10_result['aucpr'] * 100, 
        '1st Bagging': bagging_ensemble_55_result['aucpr'] * 100, '2nd Bagging': bagging_ensemble_45_result['aucpr'] * 100, 
        '3rd Bagging': bagging_ensemble_13_8_result['aucpr'] * 100, '1st Selected Bagging': selected_bagging_result['aucpr'] * 100}
df_aucpr = pd.DataFrame(data_aucpr)
print(df_aucpr.to_latex(index=False,
                  float_format="{:.2f}".format, longtable=True
                  ))
```
