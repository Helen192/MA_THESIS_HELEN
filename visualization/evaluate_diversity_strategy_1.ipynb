## 0. ensemble_top_5_best (combines top 5 best base models)

```python
ensemble_top_5_best_aucroc_scores = []
ensemble_top_5_best_aucpr_scores = []
ensemble_top_5_best_f1_scores = []
datasets_name = []
ensemble_top_5_best_time_per_instance = []
# walk through all folders and find result.csv files
for subdir, _, files in os.walk(strategy_1_ensemble_top_5_best_dir):
    if "result.csv" in files:
        file_path = os.path.join(subdir, "result.csv")
        df = pd.read_csv(file_path)

        # extract dataset name
        dataset_name = os.path.basename(os.path.dirname(subdir))
        datasets_name.append(dataset_name)

        # extract scores
        aucroc = df['aucroc'].values[0]
        ensemble_top_5_best_aucroc_scores.append(aucroc)

        aucpr = df['aucpr'].values[0]
        ensemble_top_5_best_aucpr_scores.append(aucpr)

        f1 = df['f1'].values[0]
        ensemble_top_5_best_f1_scores.append(f1)

        time_instance = df ['time-per-inst'].values[0]
        ensemble_top_5_best_time_per_instance.append(time_instance)

# compute average scores of all datasets
ensemble_top_5_best_avg_aucroc = sum(ensemble_top_5_best_aucroc_scores) / len(ensemble_top_5_best_aucroc_scores)
ensemble_top_5_best_avg_aucpr = sum(ensemble_top_5_best_aucpr_scores) / len(ensemble_top_5_best_aucpr_scores)
ensemble_top_5_best_avg_f1 = sum(ensemble_top_5_best_f1_scores) / len(ensemble_top_5_best_f1_scores)
ensemble_top_5_best_time_per_instance_avg = sum(ensemble_top_5_best_time_per_instance) / len(ensemble_top_5_best_time_per_instance)

print(f"number of datasets: {len(ensemble_top_5_best_aucroc_scores)}")
print(f"Average aucroc: {ensemble_top_5_best_avg_aucroc}")
print(f"Average aucpr: {ensemble_top_5_best_avg_aucpr}")
print(f"Average f1: {ensemble_top_5_best_avg_f1}")
print(f"Inference time per instance: {ensemble_top_5_best_time_per_instance_avg}")
```
