## 1. Ensemble_9

#### 1. ensemble_9 - strategy 1

```python
ensemble_9_aucroc_scores = []
ensemble_9_aucpr_scores = []
ensemble_9_f1_scores = []
datasets_name = []
ensemble_9_time_per_instance = []
# walk through all folders and find result.csv files
for subdir, _, files in os.walk(ensemble_9_dir):
    if "result.csv" in files:
        file_path = os.path.join(subdir, "result.csv")
        df = pd.read_csv(file_path)

        # extract dataset name
        dataset_name = os.path.basename(os.path.dirname(subdir))
        datasets_name.append(dataset_name)

        # extract scores
        aucroc = df['aucroc'].values[0]
        ensemble_9_aucroc_scores.append(aucroc)

        aucpr = df['aucpr'].values[0]
        ensemble_9_aucpr_scores.append(aucpr)

        f1 = df['f1'].values[0]
        ensemble_9_f1_scores.append(f1)

        time_instance = df ['time-per-inst'].values[0]
        ensemble_9_time_per_instance.append(time_instance)

# compute average scores of all datasets
ensemble_9_avg_aucroc = sum(ensemble_9_aucroc_scores) / len(ensemble_9_aucroc_scores)
ensemble_9_avg_aucpr = sum(ensemble_9_aucpr_scores) / len(ensemble_9_aucpr_scores)
ensemble_9_avg_f1 = sum(ensemble_9_f1_scores) / len(ensemble_9_f1_scores)
ensemble_9_time_per_instance_avg = sum(ensemble_9_time_per_instance) / len(ensemble_9_time_per_instance)

print(f"number of datasets: {len(ensemble_9_aucroc_scores)}")
print(f"Average aucroc: {ensemble_9_avg_aucroc}")
print(f"Average aucpr: {ensemble_9_avg_aucpr}")
print(f"Average f1: {ensemble_9_avg_f1}")
print(f"Inference time per instance: {ensemble_9_time_per_instance_avg}")
```
