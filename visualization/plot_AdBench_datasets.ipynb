```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

def plot_cov_ellipsoid(ax, mean, cov, nstd=2.0, num=40, linewidth=1):
    """
    Plot a covariance ellipsoid (wireframe) for a 3D Gaussian-like cloud.
    mean: (3,)
    cov:  (3,3)
    nstd: number of standard deviations for the ellipsoid radius
    """
    # Eigen-decomposition (principal axes)
    vals, vecs = np.linalg.eigh(cov)
    order = vals.argsort()[::-1]
    vals, vecs = vals[order], vecs[:, order]

    # Radii along principal axes
    radii = nstd * np.sqrt(np.maximum(vals, 1e-12))

    # Sphere parameterization
    u = np.linspace(0, 2*np.pi, num)
    v = np.linspace(0, np.pi, num)
    x = np.outer(np.cos(u), np.sin(v))
    y = np.outer(np.sin(u), np.sin(v))
    z = np.outer(np.ones_like(u), np.cos(v))

    # Scale sphere -> ellipsoid
    xyz = np.stack([x, y, z], axis=0)  # (3, num, num)
    xyz = radii[:, None, None] * xyz

    # Rotate and translate
    xyz = vecs @ xyz.reshape(3, -1)
    xyz = xyz.reshape(3, num, num)
    x_e, y_e, z_e = xyz[0] + mean[0], xyz[1] + mean[1], xyz[2] + mean[2]

    ax.plot_wireframe(x_e, y_e, z_e, rstride=3, cstride=3, linewidth=linewidth)

# -------------------------
# Your data here:
# X = ... shape (n, 3)
# y = ... shape (n,), 0 normal, 1 anomaly
# -------------------------

X = np.asarray(X)
y = np.asarray(y)

normal = X[y == 0]
anom   = X[y == 1]

# Split normal points into two clusters (N1, N2)
kmeans = KMeans(n_clusters=2, random_state=0, n_init="auto")
labels = kmeans.fit_predict(normal)

N1 = normal[labels == 0]
N2 = normal[labels == 1]

# Plot
fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection="3d")

# Normals (different markers like the example)
ax.scatter(N1[:, 0], N1[:, 1], N1[:, 2], s=12, marker="*", alpha=0.6, label="N1")
ax.scatter(N2[:, 0], N2[:, 1], N2[:, 2], s=12, marker="s", alpha=0.6, label="N2")

# Anomalies
ax.scatter(anom[:, 0], anom[:, 1], anom[:, 2], s=80, marker="o", alpha=0.9, label="A")

# Ellipsoids around N1/N2 (like the bubbles)
plot_cov_ellipsoid(ax, N1.mean(axis=0), np.cov(N1, rowvar=False), nstd=2.5)
plot_cov_ellipsoid(ax, N2.mean(axis=0), np.cov(N2, rowvar=False), nstd=2.5)

# Labels / view
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.view_init(elev=20, azim=35)
ax.legend()
plt.tight_layout()
plt.show()
```

### Project into 2d

```python
from sklearn.decomposition import PCA

import numpy as np
from matplotlib.patches import Ellipse

def plot_cov_ellipse(ax, mean, cov, nstd=2.5, **kwargs):
    """
    Draw a covariance ellipse for 2D data.
    """
    vals, vecs = np.linalg.eigh(cov)
    order = vals.argsort()[::-1]
    vals, vecs = vals[order], vecs[:, order]

    width, height = 2 * nstd * np.sqrt(np.maximum(vals, 1e-12))
    angle = np.degrees(np.arctan2(vecs[1, 0], vecs[0, 0]))

    ellipse = Ellipse(
        xy=mean,
        width=width,
        height=height,
        angle=angle,
        fill=False,
        linewidth=1.5,
        **kwargs
    )
    ax.add_patch(ellipse)


# Optional but recommended
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)

normal = X_2d[y == 0]
anom   = X_2d[y == 1]

# Cluster normals
kmeans = KMeans(n_clusters=2, random_state=0, n_init="auto")
labels = kmeans.fit_predict(normal)

N1 = normal[labels == 0]
N2 = normal[labels == 1]

fig, ax = plt.subplots(figsize=(7, 6))

ax.scatter(N1[:, 0], N1[:, 1], marker="*", s=30, alpha=0.6, label="N1")
ax.scatter(N2[:, 0], N2[:, 1], marker="s", s=30, alpha=0.6, label="N2")
ax.scatter(anom[:, 0], anom[:, 1], marker="o", s=80, alpha=0.9, label="A")

plot_cov_ellipse(ax, N1.mean(axis=0), np.cov(N1, rowvar=False))
plot_cov_ellipse(ax, N2.mean(axis=0), np.cov(N2, rowvar=False))

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_title("2D PCA projection")
ax.legend()
ax.set_aspect("equal")
plt.tight_layout()
plt.show()
```
