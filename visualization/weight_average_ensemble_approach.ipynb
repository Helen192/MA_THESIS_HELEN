```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# paths of models
model_1_dir = "../results/adbench/dataSGMM.context5000.feat20.R500.inf-fullTrue.LTFalse.gen1tr1True.reuse100.E10.step1000.bs4.lr0.001.emb256.hdim512.nhead4.nlayer4.ndevice1.numclusters20/train_seed=0.transform=quantile.feat_truncation=subsampling"
model_1_GMM_dir = "../results/adbench/dataGMM.context5000.feat20.R500.inf-fullTrue.LTFalse.gen1tr1True.reuse100.E10.step1000.bs4.lr0.001.emb256.hdim512.nhead4.nlayer4.ndevice1.numclusters20/train_seed=0.transform=quantile.feat_truncation=subsampling"
model_1_default_dir = "../results/adbench/context5000.feat20.R500.inf-fullTrue.LTFalse.gen1tr1True.reuse100.E10.step1000.bs4.lr0.001.emb256.hdim512.nhead4.nlayer4.ndevice1/train_seed=0.transform=quantile.feat_truncation=subsampling"
```

```python
weight_ensemble_1_dir = "../results/adbench/weight_average_data_prior_ensemble/train_seed=0"
weight_ensemble_2_dir = "../results/adbench/weight_average_ensemble_without_data_numclusters/train_seed=0"
weight_ensemble_3_dir = "../results/adbench/test_weight_feat_80/train_seed=0"
weight_ensemble_4_dir = "../results/adbench/test_weight_feat_70/train_seed=0"
```

## 1. ensembl_1

```python
weight_ensemble_1_aucroc_scores = []
weight_ensemble_1_aucpr_scores = []
weight_ensemble_1_f1_scores = []
datasets_name = []
weight_ensemble_1_time_per_instance = []
# walk through all folders and find result.csv files
for subdir, _, files in os.walk(weight_ensemble_1_dir):
    if "result.csv" in files:
        file_path = os.path.join(subdir, "result.csv")
        df = pd.read_csv(file_path)

        # extract dataset name
        dataset_name = os.path.basename(os.path.dirname(subdir))
        datasets_name.append(dataset_name)

        # extract scores
        aucroc = df['aucroc'].values[0]
        weight_ensemble_1_aucroc_scores.append(aucroc)

        aucpr = df['aucpr'].values[0]
        weight_ensemble_1_aucpr_scores.append(aucpr)

        f1 = df['f1'].values[0]
        weight_ensemble_1_f1_scores.append(f1)

        time_instance = df ['time-per-inst'].values[0]
        weight_ensemble_1_time_per_instance.append(time_instance)

# compute average scores of all datasets
weight_ensemble_1_avg_aucroc = sum(weight_ensemble_1_aucroc_scores) / len(weight_ensemble_1_aucroc_scores)
weight_ensemble_1_avg_aucpr = sum(weight_ensemble_1_aucpr_scores) / len(weight_ensemble_1_aucpr_scores)
weight_ensemble_1_avg_f1 = sum(weight_ensemble_1_f1_scores) / len(weight_ensemble_1_f1_scores)
weight_ensemble_1_time_per_instance_avg = sum(weight_ensemble_1_time_per_instance) / len(weight_ensemble_1_time_per_instance)

print(f"number of datasets: {len(weight_ensemble_1_aucroc_scores)}")
print(f"Average aucroc: {weight_ensemble_1_avg_aucroc}")
print(f"Average aucpr: {weight_ensemble_1_avg_aucpr}")
print(f"Average f1: {weight_ensemble_1_avg_f1}")
print(f"Inference time per instance: {weight_ensemble_1_time_per_instance_avg}")
```
